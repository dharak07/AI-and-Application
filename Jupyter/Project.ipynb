### 8. Data Preprocessing

The aim of the data preprocessing technique is to enhance the quality and uniformity of the Amazon Fine Food Reviews dataset before the exploratory analysis. Though many fields are filled, there are still some missing values in columns like ProfileName and Summary, while it is possible to have duplicate records.

1. Eliminate any repetitive records of reviews.
2. Deal with the unavailability of data in columns `ProfileName` and `Summary`.
3. Transform the column `Time` from the Unix timestamp format into a user-readable format.
4. Create features related to `Year` and `Month`.
5. Design a new feature `ReviewLength` which will be extracted from the `Text` column.
6. Compute a `HelpfulnessRatio` from the `HelpfulnessNumerator` and `HelpfulnessDenominator` values.
7. Ensure that the values of review `Score` are within the permissible range of values.

```python
# Create a copy to preserve the raw dataset
clean_df = df.copy()

# Standardize column names
clean_df.columns = clean_df.columns.str.strip()

# Remove duplicate rows
clean_df = clean_df.drop_duplicates()

# Handle missing values in important text columns
clean_df['ProfileName'] = clean_df['ProfileName'].fillna('Unknown') 
clean_df['Summary'] = clean_df['Summary'].fillna('No Summary')
clean_df['Text'] = clean_df['Text'].fillna('')

# Convert Unix timestamp into readable datetime
clean_df['ReviewDate'] = pd.to_datetime(
    clean_df['Time'], 
    unit='s'
)

# Create time-based features
clean_df['Year'] = clean_df['ReviewDate'].dt.year
clean_df['Month'] = clean_df['ReviewDate'].dt.month
clean_df['MonthName'] = clean_df['ReviewDate'].dt.month_name()

# Create review length feature
clean_df['ReviewLength'] = clean_df['Text'].str.split().str.len()

# Create helpfulness ratio
clean_df['HelpfulnessRatio'] = np.where(
    clean_df['HelpfulnessDenominator'] > 0,
    clean_df['HelpfulnessNumerator'] / 
    clean_df['HelpfulnessDenominator'],
    0
)

# Check valid rating scores
clean_df = clean_df[
    clean_df['Score'].between(1, 5)
].copy()

print("Cleaned dataset shape:", clean_df.shape)
display(clean_df.head())
```

**Observation** : The process of preprocessing borrows its definitions from the task of elimination of repeated records, filling in the blanks of missing text values, converting timestamps into readable date format, as well as extraction of useful features such as `Year`, `Month`, `ReviewLength`, and `HelpfulnessRatio`.

```python
# Verify Cleaning
print("Missing values after cleaning:")
display(clean_df.isnull().sum())

# Check duplicate rows
print("Duplicate rows after Cleaning:", clean_df.duplicated().sum())

# Check rating range
print("Minimum review score:", clean_df['Score'].min())
print("Maximum review score:", clean_df['Score'].max())

# Check review length
print("Minimum review length:", clean_df['ReviewLength'].min())
print("Maximum review length:", clean_df['ReviewLength'].max())

# Check helpfulness ratio
print("Minimum helpfulness ratio:", clean_df['HelpfulnessRatio'].min())
print("Maximum helpfulness ratio:", clean_df['HelpfulnessRatio'].max())

# Check review date range
print("Earliest review date:", clean_df['ReviewDate'].min())
print("Latest review date:", clean_df['ReviewDate'].max())

# Check rating distribution
print("\nReview Score Distribution:")
display(
    clean_df['Score']
    .value_counts()
    .sort_index()
)
```

**Observation** : The purpose of this verification is to establish whether duplicated records or missing values exist after pre-processing phase. At the same time, review scores, helpfulness rates, review length, dates, etc.

### 9. Explanatory Data Analysis

**Business Question 1** : How are customer rating scores distributed across Amazon Fine Food products?

**Importance** : When analyzing the total customer satisfaction, it is very important to analyze the ratings.

```python
# Calculate number of reviews for each rating score
score_distribution = (
    clean_df['Score']
    .value_counts()
    .sort_index()
    .reset_index()
)

score_distribution.columns = ['Score','Number_of_Reviews']

# Calculate percentage of total reviews
score_distribution['percentage'] = (
    score_distribution['Number_of_Reviews']
    / score_distribution['Number_of_Reviews'].sum()
) * 100

display(score_distribution)

# Visualisation - Bar Chart
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    score_distribution['Score'],
    score_distribution['Number_of_Reviews']
)

plt.title('Distribution of Amazon Fine Food Review Scores')
plt.xlabel('Review Score')
plt.ylabel('Number of Reviews')
plt.xticks([1, 2, 3, 4, 5])
plt.tight_layout()

plt.show()
```

**Observation** : In this analysis, the number and proportion of reviews based on all five rating scales will be compared. Hence, it would be possible to determine which ratings have been most often given, as well as whether customer reviews are mostly positive, negative, or neutral.

**Buisness Question 2** : Which products achieve the strongest ratings while also receiving a substantial number of reviews?

**Importance** : While an item might receive a perfect score of 5 based on a small number of reviews, it can’t be considered as trustworthy as another product with a 5-star rating garnered from hundreds of reviews. This leads to the point that the analysis combines `average score and review quantity` in order to find out which items, besides demonstrating high rating average, also have a decent amount of positive feedback. 

```python
# Analyse product performance using review volume and average score
product_performance = (
    clean_df.groupby('ProductId')
    .agg(
        Review_Count=('Id','count'),
        Average_Score=('Score','mean')
    )
    .reset_index()
)

# Keep products with at least 50 reviews
reliable_products = product_performance[
    product_performance['Review_Count'] >= 50
].copy()

# Identify the top 15 products
top_products = (
    reliable_products
    .sort_values(
        by=['Average_Score', 'Review_Count'],
        ascending=[False, False]
    )
    .head(15)
)

display(top_products)

# Visualisation - Scatter Plot
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(
    reliable_products['Review_Count'],
    reliable_products['Average_Score'],
    alpha=0.5
)

plt.title('Product Review Volume vs Average Rating')
plt.xlabel('Number of Reviews')
plt.ylabel('Average Review Score')
plt.ylim(1, 5.1)
plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

**Observation** : The scatter plot illustrates the different products that pass two important criteria which are the total number of reviews received and the average rating given by consumers. The position of a product on the plot indicates how high the average rating is, while its location on the horizontal axis reflects how many reviews it has received. In this way, the method eliminates the risk of poor products being classified as bestsellers on the basis of a few excellent reviews only, as it only counts products that received at least 50 reviews.

**Business Question 3** : Does review length influence how helpful customers find a review?

**Importance** :  Length of a customer review might influence the level of detail a customer may provide about his experience with the product. In relation to the helpfulness score received, it might prove useful to assess if long reviews or short ones are more beneficial. Companies may also gain from information concerning the usefulness of long reviews in assisting consumers in making their purchase decisions.

```python
# Create review length categories
clean_df['ReviewLengthGroup'] = pd.cut(
    clean_df['ReviewLength'],
    bins=[0, 25, 50, 100, 200, float('inf')],
    labels=[
        'very Short',
        'Short',
        'Medium',
        'Long',
        'Very Long'
    ]
)

# Calculate helpfulness statistics for each review length group
length_helpfulness = (
    clean_df.groupby('ReviewLengthGroup', observed=False)
    .agg(
        Number_of_Reviews=('Id', 'count'),
        Average_Helpfulness=('HelpfulnessRatio', 'mean')
    )
    .reset_index()
)

length_helpfulness['Average_Helpfulness'] *= 100

display(length_helpfulness)

# Visualisation - Bar Plot

plot_data = [
    clean_df.loc[
        clean_df['ReviewLengthGroup'] == group,
        'HelpfulnessRatio'
    ].dropna()
    for group in clean_df['ReviewLengthGroup'].cat.categories
]

plt.figure(figsize=(10, 6))

plt.boxplot(
    plot_data,
    tick_labels = clean_df['ReviewLengthGroup'].cat.categories,
    showfliers=False
)

plt.title('Review Helpfulness by Review Length')
plt.xlabel('Review Length Category')
plt.ylabel('Helpfulness Ratio')
plt.xticks(rotation=20)
plt.tight_layout()

plt.show()
```

**Observation** : The boxplot illustrates the variance of helpfulness ratios for different review lengths from shorter ones to longer ones. This would tell us whether longer reviews would mean more helpful information received by customers or not. If longer reviews have higher median helpfulness ratios, that would likely mean that consumers find helpful experience with a product helpful in their purchase decisions.

**Business Question 4** : Are extremely positive or negative reviews considered more helpful than moderate reviews?

**Importance** : Before making a purchase, customers often look to reviews for information about a product’s advantages and disadvantages. Ratings allow us to discover what customers think about the value of positive, negative, or neutral reviews in terms of usefulness. This will aid companies in realizing which type of customer feedback works best in helping other customers. 

```python
# Use only reviews that received at least one helpfulness vote
helpfulness_data = clean_df[
    clean_df['HelpfulnessDenominator'] > 0
].copy()

# Calculate helpfulness statistics for each review score
score_helpfulness = (
    helpfulness_data.groupby('Score')
    .agg(
        Number_of_Reviews = ('Id', 'count'),
        Average_Helpfulness = ('HelpfulnessRatio', 'mean'),
        Median_Helpfulness = ('HelpfulnessRatio', 'median')
    )
    .reset_index()
)

# Convert helpfulness values into percentage
score_helpfulness['Average_Helpfulness'] *= 100
score_helpfulness['Median_Helpfulness'] *= 100

display(score_helpfulness)

# Create data for violin plot
plot_data = [
    helpfulness_data.loc[
        helpfulness_data['Score'] == score,
        'HelpfulnessRatio'
    ].sample(
        n=min(
            5000,
            len(helpfulness_data[
                helpfulness_data['Score'] == score
            ])
        ),
        random_state = 42
    )
    for score in range(1, 6)
]

# Visualisation - Violin Plot
plt.figure(figsize=(9, 6))

plt.violinplot(
    plot_data,
    positions=[1, 2, 3, 4, 5],
    showmeans=True,
    showmedians=True
)

plt.title('Review Helpfulness Dustribution by Rating Score')
plt.xlabel('Review Score')
plt.ylabel('Helpfulness Ratio')
plt.xticks([1, 2, 3, 4, 5])
plt.ylim(0, 1.05)
plt.tight_layout()

plt.show()
```

**Observation** : This analysis evaluates the information contained in reviews at each one of the five rating levels. It does not include reviews with no values in the helpfulness column since being shown a zero means that no one made a determination on the helpfulness of the review. The data visual is designed to both show distribution and concentration of helpfulness rates among ratings.

**Business Question 5** : Which reviewers have the greatest influence based on review activity and helpfulness?

**Importance** : Some consumers send quite a large number of reviews while others submit fewer reviews that other consumers find very beneficial to them. Identifying users who show both great activity as well as great usefulness can point to significant players in the consumers’ review community. Such users could provide particularly useful feedback on products making it easier for a company to figure out extensive information about the opinions of consumers and their buying experience.

```python
# Calculate reviewer-level statistics
reviewer_analysis = (
    clean_df.groupby('UserId')
    .agg(
        Review_Count=('Id', 'count'),
        Average_Helpfulness=('HelpfulnessRatio', 'mean')
    )
    .reset_index()
)

# Keep reviewers with at least 10 reviews
active_reviewers = reviewer_analysis[
    reviewer_analysis['Review_Count'] >= 10
].copy()

# Create an influence score
active_reviewers['Influence_Score'] = (
    active_reviewers['Review_Count'] *
    active_reviewers['Average_Helpfulness']
)

# Select top 10 influential reviewers
top_reviewers = (
    active_reviewers
    .sort_values('Influence_Score', ascending=False)
    .head(10)
    .sort_values('Influence_Score')
)

display(top_reviewers)

# Visualisation - Lollipop Chart
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.hlines(
    y=top_reviewers['UserId'],
    xmin=0,
    xmax=top_reviewers['Influence_Score']
)

plt.plot(
    top_reviewers['Influence_Score'],
    top_reviewers['UserId'],
    'o'
)

plt.title('Top 10 Influential Reviewers')
plt.xlabel('Influence Score')
plt.ylabel('User ID')
plt.tight_layout()

plt.show()
```

**Observation** : The analysis highlights the reviewers that combine high activity in terms of reviewing with high helpfulness levels. Instead of simply ranking reviewers by the number of reviews submitted, the influence score combines the frequency of contribution as well as the perceived usefulness of those reviews for other consumers. Reviewers who are positioned on the right side of the lollipop chart are the ones with the most influence.

**Business Question 6** : How have review activity and average customer ratings changed over time?

**Importance** : Analyzing the action of reviews leads to the understanding of the way customer behavior in relation to Amazon Fine Food goods has changed. The average rate will provide the information about whether customers’ satisfaction is on stable level or has become higher or lower as compared to previous periods.

```python
time_analysis = (
    clean_df
    .set_index('ReviewDate')
    .resample('YE')
    .agg(
        Review_Count=('Id','count'),
        Average_Score=('Score','mean')
    )
    .reset_index()
)

# Extract year for display
time_analysis['Year'] = time_analysis['ReviewDate'].dt.year

display(
    time_analysis[
        ['Year','Review_Count','Average_Score']
    ]
)

# Visualisation - Dual Axis Line Chart
fig, ax1 = plt.subplots(figsize=(10, 6))

# Review volume
ax1.plot(
    time_analysis['Year'],
    time_analysis['Review_Count'],
    marker='o',
    linewidth=2
)

ax1.set_xlabel('Year')
ax1.set_ylabel('Number of Reviews')

# Second axis for average rating
ax2 = ax1.twinx()

ax2.plot(
    time_analysis['Year'],
    time_analysis['Average_Score'],
    marker='s',
    linestyle='--',
    linewidth=2
)

ax2.set_ylabel('Average Review Score')
ax2.set_ylim(1, 5)

plt.title('Review Activity and Average Rating Over Time')

fig.tight_layout()
plt.show()
```

**Observation** : The study monitors two key metrics over time: the number of reviews written by customers each year and the average ratings assigned by them. The first graph presents information of how many reviews were received for each year of interest, while the second provides evidence whether the average rating has changed over the same period of time. 

**Business Question 7** : Which products generate the most disagreement among customers?

**Importance** : According to the high average rating, it cannot be concluded that the customers approve of the product. It is seen that some products gain very high ratings as well as very low ratings meaning different people have different experiences or expectations. The analysis of the variability of the ratings helps to find products that may need special attention since mixed opinions always signify inconsistency in quality, tastes, packaging issues, or other matters regarding the product.

```python
# Calculate rating variability for each product
product_variability = (
    clean_df.groupby('ProductId')
    .agg(
        Review_Count=('Id','count'),
        Average_Score=('Score','mean'),
        Rating_Std=('Score','std')
    )
    .reset_index()
)

# Keep products with enough reviews for reliable comparison
product_variability = product_variability[
    product_variability['Review_Count'] >= 50
].copy()

# Select the 15 products with the highest rating variation
most_divisive_products = (
    product_variability
    .sort_values('Rating_Std',ascending=False)
    .head(15)
)

display(most_divisive_products)

plt.figure(figsize=(10, 7))

plt.barh(
    most_divisive_products['ProductId'],
    most_divisive_products['Rating_Std']
)

plt.title('Products with the Highest Rating Disagreement')
plt.xlabel('Rating Standard Deviation')
plt.ylabel('Product ID')

plt.gca().invert_yaxis()
plt.tight_layout()

plt.show()
```

**Observation** : The analysis points out those products, which experience the largest differences in customers’ ratings. A higher standard deviation means larger divergence in views regarding a product that receives both positive and negative ratings. The research focuses on products being rated by at least 50 users to ensure comparability of results and to lessen the effect of products being rated by too few customers.

**Business Question 8** : What combination of rating score and review length products the most helpful customer feedback?

**Importance** : This question considers multiple characteristics of reviews together, instead of looking at them separately. It is probable that if a review is useful, it does not always matter whether the rating is good or bad. The combination of score and review length is essential in understanding which reviews people find most useful.

```python
# Create review length categories if not already created
clean_df['ReviewLengthGroup'] = pd.cut(
    clean_df['ReviewLength'],
    bins=[0, 25, 50, 100, 200, float('inf')],
    labels=[
        'Very Short',
        'Short',
        'Medium',
        'Long',
        'Very Long'
    ]
)

# Use only reviews that received helpfulness votes
helpful_reviews = clean_df[
    clean_df['HelpfulnessDenominator'] > 0
].copy()

# Create pivot table of average helpfulness
helpfulness_matrix = pd.pivot_table(
    helpful_reviews,
    values='HelpfulnessRatio',
    index='Score',
    columns='ReviewLengthGroup',
    aggfunc='mean',
    observed=False
) * 100

display(helpfulness_matrix)

# Visualisation - Heatmap
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.imshow(
    helpfulness_matrix,
    aspect='auto'
)

plt.colorbar(label='Average Helpfulness (%)')

plt.xticks(
    range(len(helpfulness_matrix.columns)),
    helpfulness_matrix.columns,
    rotation=45
)

plt.yticks(
    range(len(helpfulness_matrix.index)),
    helpfulness_matrix.index
)

plt.xlabel('Review Length Category')
plt.ylabel('Review Score')
plt.title('Average Helpfulness by Rating Score and Review Length')

# Add values inside the heatmap
for i in range(len(helpfulness_matrix.index)):
    for j in range(len(helpfulness_matrix.columns)):
        value = helpfulness_matrix.iloc[i, j]

        if pd.notna(value):
            plt.text(
                j,
                i,
                f'{value:.1f}%',
                ha='center',
                va='center'
            )

plt.tight_layout()
plt.show()
```

**Observation** : In the analysis, we will compare the average helpfulness ratings and determine which combinations of review length and score are meaningful in identifying cases where specific reviews, such as “long negative reviews,” “medium-length positive reviews,” or “long 5-star reviews,” are perceived as useful by other people.
